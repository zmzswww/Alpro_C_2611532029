## 2.1 Dasar Teori
Perulangan merupakan struktur kontrol di dalam pemrograman yang bertugas mengeksekusi suatu blok kode secara berulang-ulang. Proses repetisi ini akan terus berjalan selama prasyarat atau kondisi tertentu terpenuhi, atau selama masih terdapat elemen yang belum selesai diproses. Di dalam bahasa Python, terdapat dua varian perulangan utama, yaitu for dan while. Modul perulangan for umumnya diaplikasikan apabila total pengulangan sudah dipastikan dari awal, atau saat program perlu menelusuri (iterate) suatu kumpulan data.

Di dalam ekosistem Python, perulangan for dioperasikan sebagai sebuah iterator. Variabel yang digunakan dalam perulangan akan diisi secara bertahap satu demi satu dengan elemen-elemen yang berasal dari suatu urutan data. Setelah variabel terisi, blok kode yang berada di dalamnya akan langsung dijalankan untuk masing-masing elemen tersebut. Dalam penulisan sintaksnya, penempatan tanda titik dua (:) pada ujung baris serta penerapan indentasi pada blok kode hukumnya wajib. Hal ini dikarenakan Python menggunakan sistem indentasi sebagai penanda batas cakupan blok program.

Penggunaan kata kunci in berfungsi untuk mengaitkan antara variabel perulangan dengan urutan data yang hendak ditelusuri. Secara harfiah, kata kunci ini bermakna "untuk tiap-tiap variabel yang terdapat di dalam suatu urutan". Adapun jenis data berurutan yang dapat ditelusuri meliputi list, string, tuple, set, dictionary, ataupun objek berupa range.

## 2.2 Kode Program

### 2.2.1 Program Perulangan_for1_2611532029

```python
# Buat file dengan nama perulangan_for1_NIM.py
# Buat program untuk perulangan for dalam pytohn
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2029 = int(input("Masukkan jumlah perulangan: "))

for i_2029 in range(ulang_2029):
    print(f"Perulangan ke-{i_2029}")
```

**Penjelasan :**

1. `ulang_2029 = int(input("masukkan jumlah perulangan: "))`
Baris ini berfungsi untuk memunculkan teks perintah `"masukkan jumlah perulangan: "` di layar dan menunggu user memasukkan sebuah nilai angka. Data yang diinput oleh user awalnya bertipe string, sehingga perlu dikonversi menjadi tipe bilangan bulat menggunakan fungsi `int()`, kemudian nilai tersebut dialokasikan ke dalam variabel `ulang_2029`.

2. `for i_2029 in range(ulang_2029):`
Baris ini digunakan untuk menginisialisasi proses perulangan agar berjalan sebanyak nilai yang tersimpan di dalam variabel `ulang_2029`. Fungsi `range(n)` di sini bertugas memproduksi urutan angka bernilai dari `0` hingga `n-1`, di mana nilai angka pada tiap-tiap putaran akan ditampung secara bergantian oleh variabel `i_2029`.

3. `print(f"perulangan ke-({i_2029})")`
Baris instruksi di dalam blok perulangan ini berfungsi untuk mencetak teks `"perulangan ke-"` yang disertai dengan nilai dari variabel `i_2029` di dalam tanda kurung pada setiap iterasi. Penggunaan huruf `f` (f-string) di awal tanda kutip bertujuan agar nilai variabel di dalam tanda kurung kurawal `{}` dapat langsung disisipkan ke dalam teks.

### 2.2.3 Program Perulangan_for3_2611532029

```python
# Buat file dengan nama perulangan_for3_NIM.py
# Buat program untuk perulangan for dalam pytohn
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2029 = int(input("Masukkan jumlah perulangan: "))

jumlah_2029 = 0
for i_2029 in range(1, ulang_2029+1):
    print(i_2029, end=" ")
    jumlah_2029 = jumlah_2029 + i_2029

    if i_2029 < ulang_2029:
        print(" + ", end="")
    else:
        print(" = ", jumlah_2029, end="")
print()
print("Jumlah =", jumlah_2029)
```

**Penjelasan :**

1. **Input jumlah perulangan:** Program menginstruksikan pengguna untuk memasukkan suatu nilai numerik (contohnya 5), yang kemudian dikonversi menjadi tipe data integer dan dialokasikan ke dalam variabel `ulang_2029`.

2. **Inisialisasi variabel penjumlahan:** Sebuah variabel disiapkan untuk menampung akumulasi total dari penjumlahan angka, di mana nilai awalnya diatur dari 0.

3. **Perulangan for:** Proses iterasi dieksekusi mulai dari angka 1 hingga nilai angka yang diinput sebelumnya (diberikan operasi `+ 1` dikarenakan batas berhenti pada fungsi range bersifat eksklusif atau tidak diikutsertakan).

4. **Isi pengulangan:**

	a. Perintah `print(i_2029, end=" ")` bertugas menampilkan nilai angka pada iterasi berjalan secara horizontal tanpa membuat baris baru.

	b. Operasi `jumlah_2029 = jumlah_2029 + i_2029` berfungsi mengkumulasikan angka tersebut ke dalam variabel total.
    
	c. Kondisional `if` digunakan untuk mengontrol visual output: Apabila posisi iterasi belum mencapai angka terakhir (`i < ulang`), program akan mencetak simbol `+`. Namun, jika sudah berada di angka terakhir, program akan mencetak tanda `=` diikuti dengan nilai total dari `jumlah_2029`.

5. **Output akhir:** Perintah `print()` digunakan untuk memberikan jeda pindah ke baris baru, setelah itu program akan menampilkan hasil akhir dari kalkulasi total secara keseluruhan.

### 2.2.4 Program nested_for1_2611532029

```python
# Buat file dengan nama nested_for1_NIM.py
# Buat program untuk perulangan for dalam pytohn
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2029 = int(input("Masukkan nilai batas: "))
for line_2029 in range(1, batas_2029 + 1):
    for j_2029 in range(1, (-1 * line_2029 + batas_2029) + 1):
        print(".", end="")
    print(line_2029)
```

**Penjelasan :**

1. **Input batas:** Program meminta user untuk memasukkan sebuah angka bilangan bulat (contohnya 5) yang akan merepresentasikan total baris yang akan dibentuk.

2. **Perulangan luar (Menentukan baris):** Proses iterasi ini berjalan mulai dari angka 1 hingga batas yang telah ditentukan. Tiap-tiap siklus putaran mewakili baris output yang sedang dibuat, di mana variabel `line_2029` bertindak sebagai penunjuk nomor barisnya.

3. **Perulangan dalam (Mencetak baris):** Rumus matematika `-1 * line + batas` memiliki nilai yang setara dengan `batas - line`. Melalui formula ini, kuantitas karakter titik yang ditampilkan pada tiap baris ditentukan oleh selisih antara nilai batas dengan nomor baris aktif, sehingga jumlah karakter titik akan semakin menyusut seiring bertambahnya baris ke bawah. Penggunaan argumen `end=""` menjaga agar karakter titik ditampilkan secara horizontal berdampingan.

4. **Cetak nomor baris:** Ketika proses pencetakan karakter titik pada baris tersebut sudah terpenuhi, program akan langsung menampilkan nomor barisnya dan dilanjutkan dengan instruksi pindah baris.

5. **Contoh visual output (input 5):**

    ....1<br>
    ...2<br>
    ..3<br>
    .4<br>
    5
    
    Penjelasan Pola: Ketika program mencapai baris paling akhir (`line` bernilai sama dengan `batas`), kalkulasi jumlah titik menghasilkan nilai 0. Kondisi ini membuat baris terakhir hanya menampilkan angka saja tanpa didahului titik. Hasil akhir dari kombinasi ini membentuk sebuah pola segitiga angka yang posisinya condong ke arah kiri bawah, di mana karakter titik berfungsi sebagai elemen pengisi ruang ("spasi") di sisi sebelah kiri angka.

### 2.2.5 Program nested_for2_2611532029

4. **Pindah baris:** Setelah proses pencetakan seluruh bintang pada satu baris terpenuhi, instruksi `print()` dijalankan untuk memindahkan kursor ke baris baru di bawahnya.

### 2.2.7 Program nested_for4_2611532029

```python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam pytohn
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_2029 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_2029 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2029 = tinggi_2029 
    c_2029 = a_2029
    lebar_2029 = (2 * tinggi_2029) - 2

    for i_2029 in range(1, tinggi_2029 + 1):
        b_2029 = c_2029 + 1

        for j_2029 in range(1, lebar_2029 + 1):

            # baris atas dan bawah

            if i_2029 == 1 or i_2029 == tinggi_2029:
                if j_2029 == 1 or j_2029 == lebar_2029:
                    print("#", end= "")
                else:
                    print("=", end= "")
            # Baris isi
            else:
                if j_2029 == 1 or j_2029 == lebar_2029:
                    print("|", end="")
                else:
                    if j_2029 == c_2029:
                        print("<", end="")
                    elif j_2029 == b_2029:
                        print(">", end="")
                    elif j_2029 == (lebar_2029 - c_2029):
                        print("<", end="")
                    elif j_2029 == (lebar_2029 - c_2029 + 1):
                        print(">", end="")
                    elif j_2029 > b_2029 and j_2029 < (lebar_2029 - c_2029):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # logika asli java

        a_2029 -= 2

        if a_2029 <= 0:
            c_2029 = (-a_2029) + 2
        else:
            c_2029 = a_2029
```

**Penjelasan :**

1. **Input dan validasi:** Program menerima input berupa nilai tinggi pola dari pengguna. Melalui sistem percabangan, jika angka yang dimasukkan adalah bilangan ganjil, program akan mengeluarkan pesan kesalahan (error). Sebaliknya, jika input berupa bilangan genap, instruksi akan diteruskan ke blok `else` untuk memproses pola.

2. **Inisialisasi variabel:** Variabel `a_2029` dan `c_2029` diatur dengan nilai awal yang sama dengan tinggi pola, di mana `c` merepresentasikan posisi koordinat kolom untuk karakter `<` di bagian kiri. Sementara itu, dimensi horizontal pola dihitung lewat rumus `lebar_2029 = 2 * tinggi - 2` (sebagai contoh, input tinggi 10 akan menghasilkan lebar pola sebesar 18 kolom).

3. **Perulangan luar (Baris):** Iterasi ini bertugas mengontrol jalannya baris mulai dari angka 1 hingga nilai tinggi. Pada tiap-tiap baris, variabel `b` dikalkulasikan lewat rumus `c + 1` untuk menetapkan posisi karakter `>` yang menjadi pasangan dari karakter `<`.

4. **Perulangan dalam (Kolom):** Proses iterasi kolom `j` bergerak dari angka 1 hingga nilai lebar untuk menentukan karakter spesifik yang akan ditampilkan pada posisi tersebut:

	a. **Baris paling awal dan paling akhir:** Karakter `#` dicetak pada kolom sudut (pojok), sedangkan posisi kolom di antaranya diisi dengan karakter `=`.

	b. **Baris bagian tengah (isi):**

		1). Kolom pertama dan kolom terakhir menampilkan karakter `|`.

		2). Kolom indeks `c` serta kolom `lebar - c` menampilkan karakter `<`.

		3). Kolom indeks `b` serta kolom `lebar - c + 1` menampilkan karakter `>`.

		4). Area kolom yang berada di dalam rentang pasangan `<>` (baik di sisi kiri maupun kanan) akan diisi dengan karakter titik `.`.

		5). Posisi kolom di luar kondisi tersebut akan dicetak sebagai karakter spasi kosong.

5. **Pindah baris dan pembaruan nilai `c`:** Setelah proses penyusunan satu baris selesai, instruksi `print()` dijalankan untuk beralih ke baris baru. Selanjutnya, nilai variabel `a` dikurangi dengan angka 2, lalu variabel `c` disesuaikan dengan kondisi berikut:

	• Selama nilai `a > 0`, maka `c = a`. Kondisi ini membuat nilai `c` mengalami penurunan bertahap (misalnya dari 8, 6, 4, ke 2).

```python
# === PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===

# 1. Mengambil masukan dinamis dari pengguna
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
n_2029 = int(input("Masukkan ukuran skala jam pasir (N): "))

# 2. Bingkai Pembatas Horizontal Atas
print("#", end="")
for i_2029 in range(4 * n_2029 + 5):
    print("=", end="")
print("#")

# 3. Fase 1: Jam Pasir Atas (Baris N turun s.d. 1)
for baris_2029 in range(n_2029, 0, -1):
    print("| ", end="")  # Sisi kiri dibatasi garis tegak (|) dan satu spasi padding
    
    # Spasi penyeimbang kiri: 2 * (N - baris)
    for spasi_2029 in range(2 * (n_2029 - baris_2029)):
        print(" ", end="")
        
    # Deret angka mundur dari baris ke 1, dipisahkan spasi
    for angka_2029 in range(baris_2029, 0, -1):
        print(angka_2029, end=" ")
        
    # Poros kristal tengah
    print("<*>", end="")
    
    # Deret angka maju dari 1 ke baris, diawali spasi
    for angka_2029 in range(1, baris_2029 + 1):
        print(" ", end="")
        print(angka_2029, end="")
        
    # Spasi penyeimbang kanan: 2 * (N - baris)
    for spasi_2029 in range(2 * (n_2029 - baris_2029)):
        print(" ", end="")
        
    print(" |")  # Satu spasi padding dan garis tegak (|) di sisi kanan

# 4. Fase 2: Poros Titik Pusat Jam Pasir (Singularity)
print("|", end="")  # Karakter garis tegak pembatas (|) di kiri
for spasi_2029 in range(2 * n_2029 + 1):  # Spasi penyeimbang kiri: (2 * N + 1)
    print(" ", end="")
print("<*>", end="")  # Karakter poros kristal tunggal
for spasi_2029 in range(2 * n_2029 + 1):  # Spasi penyeimbang kanan: (2 * N + 1)
    print(" ", end="")
print("|")  # Karakter garis tegak pembatas (|) di kanan

# 5. Fase 3: Jam Pasir Bawah (Baris 1 naik s.d. N)
for baris_2029 in range(1, n_2029 + 1):
    print("| ", end="")  # Sisi kiri dibatasi garis tegak (|) dan satu spasi padding
    
    # Spasi penyeimbang kiri: 2 * (N - baris)
    for spasi_2029 in range(2 * (n_2029 - baris_2029)):
        print(" ", end="")
        
    # Deret angka mundur dari baris ke 1, dipisahkan spasi
    for angka_2029 in range(baris_2029, 0, -1):
        print(angka_2029, end=" ")
        
    # Poros kristal tengah
    print("<*>", end="")
    
    # Deret angka maju dari 1 ke baris, diawali spasi
    for angka_2029 in range(1, baris_2029 + 1):
        print(" ", end="")
        print(angka_2029, end="")
        
    # Spasi penyeimbang kanan: 2 * (N - baris)
    for spasi_2029 in range(2 * (n_2029 - baris_2029)):
        print(" ", end="")
        
    print(" |")  # Satu spasi padding dan garis tegak (|) di sisi kanan

# 6. Bingkai Pembatas Horizontal Bawah
print("#", end="")
for i_2029 in range(4 * n_2029 + 5):
    print("=", end="")
print("#")
```

Logika Nested Loop: Sesuai ketentuan rumus \((4 \times N + 5)\), program mencetak karakter # di awal, lalu masuk ke perulangan for untuk mencetak karakter = satu per satu sebanyak rumus tersebut tanpa membuat baris baru (end=""), dan ditutup dengan karakter # di akhir diikuti perpindahan baris otomatis.

3. **Fase 1: Jam Pasir Bagian Atas (Reduksi Angka Menurun)**

    `for baris_2029 in range(n_2029, 0, -1):`

Perulangan utama bergerak mundur dari nilai skala tertinggi N turun sampai angka 1 (range(n_2029, 0, -1)). Setiap iterasi merepresentasikan satu baris jam pasir bagian atas.
