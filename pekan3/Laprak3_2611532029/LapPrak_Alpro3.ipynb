## 2.1 Dasar Teori
Fokus utama dari praktikum kali ini adalah menelaah ragam jenis operator pada Python, mengenali kaidah sintaksis dasar, serta memahami logika pemrosesan awal dalam pemrograman Python. Demi kelancaran aktivitas di laboratorium, dirumuskan beberapa pilar teori penunjang sebagai berikut:

### 2.1.1 Operator Aritmatika.
Perkakas ini dioperasikan untuk menuntaskan perhitungan matematika fundamental, mulai dari operasi pertambahan, pengurangan, perkalian, hingga kalkulasi perpangkatan.

| Operator | Deskripsi | Contoh |
| :---: | :--- | :---: |
| `+` | Penjumlahan | `a + b` |
| `-` | Pengurangan | `a - b` |
| `*` | Perkalian | `a * b` |
| `/` | Pembagian (hasil berupa *float*) | `a / b` |
| `%` | Sisa Bagi (*Modulus*) | `a % b` |
| `**` | Pangkat (*Exponentiation*) | `a ** b` |
| `//` | Pembagian Bulat (*Floor Division*) | `a // b` |

### 3.1.3 Operator Penugasan (Assignment).
Fungsi utama dari perkakas ini adalah memasukkan atau memperbarui data di dalam suatu variabel. Di samping model penugasan standar (=), Python juga dibekali dengan variasi penugasan gabungan (augmented) guna meringkas sintaksis perhitungan aritmatika.

| Operator | Contoh | Ekspresi Setara | Penjelasan |
| :---: | :---: | :---: | :--- |
| `=` | `x = 8` | `x = 8` | Mengisi nilai `8` ke dalam variabel `x` |
| `+=` | `x += 5` | `x = x + 5` | Menambahkan nilai `5` ke nilai `x` saat ini |
| `-=` | `x -= 3` | `x = x - 3` | Mengurangi nilai `x` saat ini dengan `3` |
| `*=` | `x *= 7` | `x = x * 7` | Mengalikan nilai `x` saat ini dengan `7` |
| `/=` | `x /= 3` | `x = x / 3` | Membagi nilai `x` saat ini dengan `3` (hasil *float*) |

### 2.1.3 Operator Bitwise.
Perkakas bitwise dirancang untuk memproses data berjenis bilangan bulat (integer) langsung pada tingkatan bit atau biner dengan cara menyejajarkan bit demi bit.

| Operator | Nama Operator | Deskripsi |
| :---: | :--- | :--- |
| `&` | **AND** | Bernilai `1` jika kedua bit bernilai `1` |
| `\|` | **OR** | Bernilai `1` jika salah satu bit bernilai `1` |
| `^` | **XOR** | Bernilai `1` jika bit yang dibandingkan berbeda |
| `~` | **NOT** | Membalik nilai bit (`0` menjadi `1`, dan sebaliknya) |
| `<<` | **Left Shift** | Menggeser bit ke kiri |
| `>>` | **Right Shift** | Menggeser bit ke kanan |

### 2.1.4 Operator Logika.
Instruksi ini dipakai untuk menguji keterkaitan antara dua parameter logis atau ekspresi kondisi. Output dari kalkulasi logika ini berupa representasi kebenaran (Boolean), yakni True atau False.

| Operator | Deskripsi | Contoh | Hasil |
| :---: | :--- | :---: | :---: |
| `and` | Bernilai `True` jika kedua kondisi bernilai `True` | `(5 > 2) and (3 < 8)` | `True` |
| `or` | Bernilai `True` jika salah satu kondisi bernilai `True` | `(5 > 2) or (3 > 8)` | `True` |
| `not` | Membalikkan nilai logika (`True` jadi `False`, dan sebaliknya) | `not(5 > 2)` | `False` |

### 2.1.5 Operator Keanggotaan dan Operator Identitas.
#### 2.1.5.1 Operator Keanggotaan.
Dipakai untuk mendeteksi apakah suatu elemen spesifik terdaftar di dalam susunan data kolektif (misalnya objek list, tuple, string, atau dictionary).

| Operator | Deskripsi | Contoh Ekspresi | Hasil (`a = [1, 2, 3]`) |
| :---: | :--- | :---: | :---: |
| `in` | Bernilai `True` jika nilai ditemukan di dalam urutan/koleksi | `1 in a` | `True` |
| `not in` | Bernilai `True` jika nilai tidak ditemukan di dalam urutan/koleksi | `5 not in a` | `True` |

#### 2.1.5.2 Operator Identitas.
Dipakai khusus untuk membandingkan kesamaan alamat fisik memori dari dua buah objek, jadi bukan sekadar menakar kesamaan nilai datanya saja.

| Operator | Deskripsi | Contoh Ekspresi | Hasil (`x = [1, 2]`, `y = [1, 2]`, `z = x`) |
| :---: | :--- | :---: | :---: |
| `is` | Bernilai `True` jika kedua variabel merujuk pada objek yang sama di memori | `x is z` | `True` |
| `is not` | Bernilai `True` jika kedua variabel merujuk pada objek yang berbeda | `x is not y` | `True` |


### 2.1.6 Operator Perbandingan.
Instruksi perbandingan dioperasikan untuk menakar hubungan antara dua buah nilai. Evaluasi dari perbandingan ini akan selalu menghasilkan tipe Boolean (True atau False).

| Operator | Deskripsi | Contoh | Hasil (jika `a = 5`, `b = 3`) |
| :---: | :--- | :---: | :---: |
| `==` | Sama dengan | `a == b` | `False` |
| `!=` | Tidak sama dengan | `a != b` | `True` |
| `>` | Lebih besar dari | `a > b` | `True` |
| `<` | Lebih kecil dari | `a < b` | `False` |
| `>=` | Lebih besar dari atau sama dengan | `a >= b` | `True` |
| `<=` | Lebih kecil dari atau sama dengan | `a <= b` | `False` |

### 2.2.1 Program "Aritmatika_2611532029"
Pada bagian ini, program digunakan untuk mengimplementasikan penggunaan operator aritmatika pada Python, seperti penjumlahan (+), pengurangan (-), pembagian (/), pembagian bulat (//), sisa bagi (%), dan perpangkatan (**).

```python
angka1_2029 = int(input("Input angka-1= "))
angka2_2029 = int(input("Input angka-2= "))

#Penjumlahan
hasil_2029 = angka1_2029 + angka2_2029
print("\nOperator Penjumlahan")
print("Hasil = ", hasil_2029)

#Pengurangan
hasil_2029 = angka1_2029 - angka2_2029
print("\nOperator Pengurangan")
print("Hasil = ", hasil_2029)

#Perkalian
hasil_2029 = angka1_2029 * angka2_2029
print("\nOperator Perkalian")
print("Hasil = ", hasil_2029)

#Pembagian, Pembagian Bulat, Sisa Bagi
if angka2_2029 != 0:
    hasil_2029 = angka1_2029 / angka2_2029
    print("\nOperator Pembagian")
    print("Hasil = ", hasil_2029)

    hasil_2029 = angka1_2029 // angka2_2029
    print("\nOperator Pembagian Bulat")
    print("Hasil = ", hasil_2029)

    hasil_2029 = angka1_2029 % angka2_2029
    print("\nOperator Sisa Bagi")
    print("Hasil = ", hasil_2029)
else:
    print("Angka kedua tidak boleh bernilai 0.")

#Pangkat
hasil_2029 = angka1_2029 ** angka2_2029
print("\nOperator Pangkat")
print("Hasil = ", hasil_2029)
```

**Ulasan Logika:**

1. **Penjumlahan (`+`):** Bertugas menggabungkan total nilai yang ada pada variabel `angka1_2029` dengan `angka2_2029`.

2. **Pengurangan (`-`):** Bertugas menghitung hasil selisih dari variabel `angka1_2029` yang dikurangi oleh `angka2_2029`.

3. **Pembagian (`/`):** Bertugas membagi komponen `angka1_2029` dengan `angka2_2029` yang secara berkala membuahkan data pecahan (float).

4. **Pembagian Bulat (`//`):** Bertugas mengalkulasi hasil bagi antara `angka1_2029` dan `angka2_2029` kemudian membuang nilai desimalnya ke bawah (floor division).

5. **Sisa Bagi (`%`):** Bertugas melacak sisa numerik pasca-operasi pembagian antara komponen `angka1_2029` dengan `angka2_2029`.

6. **Perpangkatan (`**`):** Bertugas mengalkulasi hasil dari variabel `angka1_2029` yang dipangkatkan sebesar nilai `angka2_2029`.

### 2.2.2 Program "Assignment_2611532029"
Pada bagian ini, program digunakan untuk mengimplementasikan penggunaan operator penugasan (assignment) pada Python, meliputi penugasan dasar (=) serta operator penugasan gabungan seperti penambahan (+=), pengurangan (-=), perkalian (*=), pembagian (/=), pembagian bulat (//=), sisa bagi (%=), dan perpangkatan (**=).

```python
angka1_2029 = int(input("Input angka ke-1 : "))
angka2_2029 = int(input("Input angka ke-2 : "))

print("\nNilai awal angka1 =",angka1_2029)
print("\nNilai awal angka2 =",angka2_2029)

#Assignment biasa
hasil = angka1_2029
print("\nAssignment biasa(=)")
print("Hasil =",hasil)

#Assignment penambahan
hasil = angka1_2029
hasil += angka2_2029
print("\nAssignment penambahan(+=)")
print("Hasil =",hasil)

#Assignment pengurangan
hasil = angka1_2029
hasil -= angka2_2029
print("\nAssignment pengurangan(-=)")
print("Hasil =",hasil)

#Assignment perkalian 
hasil = angka1_2029
hasil *= angka2_2029
print("\nAssignment perkalian(*=)")
print("Hasil =",hasil)

#Assignment pembagian, pembagian bulat, dan sisa bagi 
if angka2_2029 != 0:
    hasil = angka1_2029
    hasil /= angka2_2029
    print("\nAssignment pembagian(/=)")
    print("Hasil =",hasil)
    #Operator tambahan
    #Pembagian bulat
    hasil = angka1_2029
    hasil //= angka2_2029
    print("\nAssignment pembagian(//=)")
    print("Hasil =",hasil)
    #Sisa bagi
    hasil = angka1_2029
    hasil %= angka2_2029
    print("\nAssignment pembagian(%=)")
    print("Hasil =",hasil)
else:
    print("\nPembagian tidak dapat")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil = angka1_2029
hasil **= angka2_2029
print("\nAssignment perpangkatan (**=)")
print("Hasil =",hasil)
```

**Ulasan Logika:**

1. **Assignment Biasa (`=`):** Bertindak mengalokasikan data milik `angka1_2029` langsung ke dalam variabel hasil.

2. **Assignment Penambahan (`+=`):** Berfungsi memperbarui variabel hasil dengan menambahkan nilai `angka2_2029` ke dalamnya.

3. **Assignment Pengurangan (`-=`):** Berfungsi memperbarui nilai hasil dengan menguranginya sebesar angka pada `angka2_2029`.

4. **Assignment Perkalian (`*=`):** Berfungsi memperbarui nilai hasil dengan mengalikannya secara langsung dengan `angka2_2029`.

5. **Assignment Pembagian (`/=`):** Berfungsi memperbarui data hasil melalui pembagian dengan `angka2_2029`.

6. **Assignment Pembagian Bulat (`//=`):** Berfungsi membagi data hasil dengan `angka2_2029` lalu menerapkan pembulatan ke bawah secara otomatis.

7. **Assignment Sisa Bagi (`%=`):** Berfungsi mengisi variabel hasil dengan angka sisa bagi dari pembagian dengan `angka2_2029`.

8. **Assignment Perpangkatan (`**=`):** Berfungsi mengubah isi data hasil menjadi hasil pangkat dari nilai `angka2_2029`.

### 2.2.3 Program "Bitwise_2611532029"
Pada bagian ini, program digunakan untuk mengimplementasikan penggunaan operator bitwise pada Python untuk melakukan operasi pada tingkat biner (bit), meliputi operasi AND (&), OR (|), XOR (^), NOT (~), serta pergeseran bit yaitu Shift Left (<<) dan Shift Right (>>), lengkap dengan penampil format biner menggunakan fungsi bin() dan format().

```python
print("\n===================")
print("3. OPERATOR BITWISE")
print("===================")

angka1_2029 = int(input("Masukkan angka bitwise-1: "))
angka2_2029 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1_2029 =", angka1_2029, "| biner", bin(angka1_2029))
print("angka2_2029 =", angka2_2029, "| biner", bin(angka2_2029))

# Bitwise AND
hasil_2029 = angka1_2029 & angka2_2029
print("\nBitwise AND (&)")
print(angka1_2029, "&", angka2_2029, "=", hasil_2029)
print("Biner hasil =", bin(hasil_2029))
print("Biner hasil (8 bit) =", format(hasil_2029, "08b"))

# Bitwise OR
hasil_2029 = angka1_2029 | angka2_2029
print("\nBitwise OR (|)")
print(angka1_2029, "|", angka2_2029, "=", hasil_2029)
print("Biner hasil =", bin(hasil_2029))
print("Biner hasil (8 bit) =", format(hasil_2029, "08b"))

# Bitwise XOR
hasil_2029 = angka1_2029 ^ angka2_2029
print("\nBitwise XOR (^)")
print(angka1_2029, "^", angka2_2029, "=", hasil_2029)
print("Biner hasil =", bin(hasil_2029))
print("Biner hasil (8 bit) =", format(hasil_2029, "08b"))

# Bitwise NOT
hasil_2029 = ~angka1_2029
print("\nBitwise NOT (~)")
print("~", angka1_2029, "=", hasil_2029)
print("Biner hasil =", bin(hasil_2029))
print("Biner hasil (8 bit) =", format(hasil_2029, "08b"))

# Bitwise geser kiri
jumlah_geser_2029 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_2029 = angka1_2029 << jumlah_geser_2029
print("\nBitwise geser kiri (<<)")
print(angka1_2029, "<<", jumlah_geser_2029, "=", hasil_2029)
print("Biner hasil =", bin(hasil_2029))
print("Biner hasil (8 bit) =", format(hasil_2029, "08b"))

# Bitwise geser kanan
hasil_2029 = angka1_2029 >> jumlah_geser_2029
print("\nBitwise geser kanan (>>)")
print(angka1_2029, ">>", jumlah_geser_2029, "=", hasil_2029)
print("Biner hasil =", bin(hasil_2029))
print("Biner hasil (8 bit) =", format(hasil_2029, "08b"))
```

**Ulasan Logika:**

1. **Konversi Biner (`bin()` dan `format()`):** Utilitas `bin()` dipakai untuk memunculkan representasi biner terenkode awal, sementara instruksi `format(..., "08b")` difungsikan guna mencetak susunan biner rapi berukuran 8-bit.

2. **Bitwise AND (`&`):** Melakukan komparasi biner sejajar dan menetapkan hasil `1` apabila sepasang bit yang diuji sama-sama berstatus `1`.

3. **Bitwise OR (`|`):** Melakukan komparasi biner sejajar dan menetapkan hasil `1` jika salah satu atau seluruh bit berstatus `1`.

4. **Bitwise XOR (`^`):** Melakukan komparasi biner sejajar dan menetapkan hasil `1` khusus jika sepasang bit tersebut tidak seragam nilainya.

5. **Bitwise NOT (`~`):** Mengeksekusi penukaran bit secara total pada variabel `angka1_2029` (`0` ditukar `1` dan sebaliknya).

6. **Bitwise Geser Kiri (`<<`):** Menggeser deretan bit variabel angka`1`_2029 menuju arah kiri sejauh parameter `jumlah_geser_2029`.

7. **Bitwise Geser Kanan (`>>`):** Menggeser deretan bit variabel angka`1`_2029 menuju arah kanan sejauh parameter `jumlah_geser_2029`.

### 2.2.4 Program "Logika_2611532029"
Pada bagian ini, program digunakan untuk mengimplementasikan penggunaan operator logika pada Python untuk mengevaluasi ekspresi nilai kebenaran (boolean), meliputi Konjungsi (and), Disjungsi (or), Negasi (not), serta Disjungsi Eksklusif (XOR) yang dibentuk menggunakan operator perbandingan (!=).

```python
a1_2029 = input("Input nilai boolean-1 (True/false): ").strip().lower()== "true"
a2_2029 = input("Input nilai boolean-2 (True/false): ").strip().lower()== "true"

print("\nA1", a1_2029)
print("\nA2", a2_2029)

# Konjungsi : bernilai True jika keduanya True
hasil = a1_2029 and a2_2029
print("\nKonjungsi(AND)")
print("A1 and A2 =", hasil)

# Disjungsi : bernilai True jika salah satunya True
hasil = a1_2029 or a2_2029
print("\nDisjungsi(OR)")
print("A1 or A2 =", hasil)

# Negasi a1_2029 : membalik nilai a1_2029
hasil = not a1_2029
print("\nNegasi A1(NOT)")
print("not A1 =", hasil)

# Negasi a2_2029 : membalik nilai a2_2029
hasil = not a2_2029
print("\nNegasi A2(NOT)")
print("not A2 =", hasil)

#XOR: bernilai True jika kedua nilai berbeda
hasil = a1_2029 != a2_2029
print("\nDisjungsi Ekslusif (XOR)")
print("A1 XOR A2=",hasil)
```

**Ulasan Logika:**

1. **Input Boolean (`.strip().lower() == "true"`):** Membaca kiriman string pengguna, membersihkan spasi tak berguna lewat `.strip()`, meragamkan huruf kecil melalui `.lower()`, kemudian menvalidasinya dengan string `"true"` untuk memicu konversi tipe data boolean murni.

2. **Konjungsi / AND (`and`):** Operasi gabungan yang membuahkan hasil True hanya jika variabel `a1` beserta `a2` berstatus True secara berbarengan.

3. **Disjungsi / OR (`or`):** Operasi gabungan yang membuahkan hasil True apabila salah satu atau kedua variabel (`a1` atau `a2`) berstatus True.

4. **Negasi / NOT (`not`):** Operasi pembalikan status logika (jika data awal bernilai True maka keluaran menjadi False, begitupun sebaliknya).

5. **Disjungsi Eksklusif / XOR (`!=`):** Mengoptimalkan instruksi perbandingan `!=` untuk merekayasa pola gerbang logika XOR, yakni menghasilkan True hanya jika `a1` dan `a2` memiliki nilai kebenaran yang kontradiktif.

### 2.2.5 Program "Nama_lainnya_2611532029"
Pada bagian ini, program digunakan untuk mengimplementasikan penggunaan operator keanggotaan (membership) meliputi IN (in) dan NOT IN (not in) untuk mengecek keberadaan suatu elemen dalam list, serta operator identitas (identity) meliputi IS (is) dan IS NOT (is not) untuk membandingkan referensi alamat memori dari objek yang dibuat secara langsung maupun melalui metode pengkopian (.copy()).

```python
print("=========================")
print("1. OPERATOR KEANGGOTAAN")
print("=========================")

#Input beberarpa data yang ingin dipisahkan dengan koma
input_data_2029 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_2029 = [int(angka.strip()) for angka in input_data_2029 .split (",")]

nilai_dicari_2029 = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil_2029 = nilai_dicari_2029 in data_2029
print("\nOperator keanggotaan IN")
print(nilai_dicari_2029,"in",data_2029,"=",hasil_2029)

#Operator not in
hasil_2029 = nilai_dicari_2029  not in data_2029
print("\nOperator keanggotaan not IN")
print(nilai_dicari_2029,"not in",data_2029,"=",hasil_2029)

print("=========================")
print("2. OPERATOR IDENTITAS")
print("=========================")

# objek1_2029 menggunakan list dari input pengguna
objek1_2029 = data_2029

# objek2_2029 merujuk pada objek yang sama dengan objek1_2029
objek2_2029 = objek1_2029

# objek3_2029 memiliki isi sama, tetapi merupakan objek baru
objek3_2029 = data_2029.copy()

print("objek1_2029 =", objek1_2029)
print("objek2_2029 =", objek2_2029)
print("objek3_2029 =", objek3_2029)

# Operator is
hasil_2029 = objek1_2029 is objek2_2029
print("\nOperator identitas IS")
print("objek1_2029 is objek2_2029 =", hasil_2029)

# Operator is not
hasil_2029 = objek1_2029 is not objek3_2029
print("\nOperator identitas IS NOT")
print("objek1_2029 is not objek3_2029 =", hasil_2029)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1_2029 is objek3_2029 =", objek1_2029 is objek3_2029)
print("objek1_2029 == objek3_2029 =", objek1_2029 == objek3_2029)
```

**Ulasan Logika:**

1. **Pengolahan Input List (`[int(...) for ...]`):** Mengonversi untaian digit masukan pengguna yang terpisah tanda koma menjadi sekumpulan objek list bertipe numerik bulat.

2. **Operator Keanggotaan IN (`in`):** Menelusuri keberadaan data `nilai_dicari_2029` di dalam objek list `data_2029` dan melontarkan True jika data terbukti eksis.

3. **Operator Keanggotaan NOT IN (`not in`):** Menelusuri ketidakberadaan data `nilai_dicari_2029` di dalam list `data_2029` dan melontarkan True jika elemen tersebut memang absen dari list.

4. **Inisialisasi Referensi Alokasi Objek:**

   a. `objek2_2029 = objek1_2029`: Memetakan variabel `objek2_2029` agar langsung merujuk pada alamat blok memori yang digunakan oleh `objek1_2029`.

   b. `objek3_2029 = data_2029``.copy()`: Menyalin muatan list lewat fungsi `.copy()`, menghasilkan variabel `objek3_2029` dengan isi yang identik namun menempati alamat memori yang baru.

5. **Operator Identitas IS (`is`):** Menakar kelayakan apakah sepasang variabel bertumpu pada lokasi memori yang sama persis. Memicu nilai True pada pengujian `objek1_2029 is objek2_2029`.

6. **Operator Identitas IS NOT (`is not`):** Menakar kelayakan apakah sepasang variabel berdiri di alamat memori yang berlainan. Memicu nilai True pada pengujian `objek1_2029 is not objek3_2029`.

7. **Komparasi `is` dan `==`:** Memberikan pembuktian bahwa `objek1_2029 is objek3_2029` menghasilkan False akibat ketidakcocokan alamat memori fisik, sedangkan `objek1_2029 == objek3_2029` membuahkan True karena isi datanya setara.

### 2.2.6 Program "Perbandingan_2611532029"
Pada bagian ini, program digunakan untuk mengimplementasikan penggunaan operator perbandingan (relational) pada Python untuk membandingkan dua nilai, meliputi Lebih Besar (>), Lebih Kecil (<), Lebih Besar dari atau Sama Dengan (>=), Lebih Kecil dari atau Sama Dengan (<=), Sama Dengan (==), Tidak Sama Dengan (!=), serta penerapan Perbandingan Berantai (chained comparison).

```python
angka1_2029 = int(input("Input angka ke-1 : "))
angka2_2029 = int(input("Input angka ke-2 : "))

#Lebih besar dari 
hasil_2029 = angka1_2029 > angka2_2029
print("\nOperator Lebih Besar")
print("Hasil = ",hasil_2029)

#Lebih kecil dari 
hasil_2029 = angka1_2029 < angka2_2029
print("\nLebih kecil dari ")
print("Hasil = ",hasil_2029)

#Lebih besar dari atau sama dengan 
hasil_2029 = angka1_2029 >= angka2_2029
print("\nOperator Lebih besar dari atau sama dengan")
print("angka1 >= angka2 = ",hasil_2029)

#Lebih kecil dari atau sama dengan
hasil_2029 = angka1_2029 <= angka2_2029
print("\nOperator Lebih kecil dari atau sama dengan")
print("angka1 <= angka2 = ",hasil_2029)

#Sama dengan 
hasil_2029 = angka1_2029 == angka2_2029
print("\nOperator Sama dengan")
print("angka1 == angka2 = ",hasil_2029)

#Tidak sama dengan 
hasil_2029 = angka1_2029 != angka2_2029
print("\nOperator Tidak sama dengan")
print("Hasil = ",hasil_2029)


#Tambahan: Perbandingan berantai dalam python
hasil_2029 = 0 < angka1_2029 < 100
print("\nOperator Perbandingan berantai")
print("0 < angka1 < 100 ",hasil_2029)

hasil_2029 = 0 < angka2_2029 < 100
print("0 < angka2 < 100",hasil_2029)
```

**Ulasan Logika:**

2. **Operator Lebih Kecil (`<`):** Menguji status kelayakan apakah variabel `angka1_2029` punya nilai di bawah `angka2_2029`.

3. **Operator Lebih Besar dari atau Sama Dengan (`>=`):** Memastikan apakah data `angka1_2029` bernilai di atas atau minimal mengimbangi angka pada `angka2_2029`.

4. **Operator Lebih Kecil dari atau Sama Dengan (`<=`):** Memastikan apakah data `angka1_2029` bernilai di bawah atau maksimal mengimbangi angka pada `angka2_2029`.

5. **Operator Sama Dengan (`==`):** Memvalidasi kesetaraan mutlak apakah variabel `angka1_2029` bernilai identik dengan `angka2_2029`.

6. **Operator Tidak Sama Dengan (`!=`):** Memvalidasi apakah variabel `angka1_2029` memiliki ketidakcocokan nilai dengan `angka2_2029`.

7. **Perbandingan Berantai (`0 < `angka1_2029` < 100`):** Optimalisasi sintaksis Python untuk menvalidasi sepasang prasyarat batas jangkauan numerik secara stimultan dalam satu baris kode tunggal.

```python
# === SISTEM SIMULASI TRANSAKSI DAN VALIDASI AKSES TOKO ===
# Menggunakan akhiran NIM 2029 untuk penamaan variabel

# 1. INPUT DATA PELANGGAN & TRANSAKSI
print("=== SISTEM TRANSAKSI TOKO ===")
nama_2029 = input("Masukkan Nama Pelanggan : ")
status_2029 = input("Masukkan Status Pelanggan (member/nonmember) : ").lower()
total_belanja_2029 = float(input("Masukkan Total Belanja : "))
jumlah_barang_2029 = int(input("Masukkan Jumlah Barang : "))
kode_promo_2029 = input("Masukkan Kode Promo : ").upper()

# Daftar promo yang tersedia untuk Operator Membership
daftar_promo_2029 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]


# 2. OPERATOR PERBANDINGAN
# Memeriksa syarat minimum belanja dan jumlah barang
syarat_belanja_2029 = total_belanja_2029 >= 200000
syarat_barang_2029 = jumlah_barang_2029 >= 3
is_member_2029 = status_2029 == "member"


# 3. OPERATOR LOGIKA & OPERATOR MEMBERSHIP
# Operator Keanggotaan (Membership) untuk cek kode promo
promo_tersedia_2029 = kode_promo_2029 in daftar_promo_2029
promo_invalid_2029 = kode_promo_2029 not in daftar_promo_2029

# Operator Logika (and, or, not)
# Mendapat diskon member jika statusnya member DAN belanja >= 200.000
get_diskon_2029 = is_member_2029 and syarat_belanja_2029
# Mendapat promo tambahan jika jumlah barang cukup ATAU kode promo tersedia, dan syarat belanja tidak gagal
get_promo_2029 = (syarat_barang_2029 or promo_tersedia_2029) and (not promo_invalid_2029)


# 4. OPERATOR ARITMATIKA
# Menghitung besaran diskon, total akhir, rata-rata, dan sisa pembagian %
besaran_diskon_2029 = 0.0
if get_diskon_2029:
    besaran_diskon_2029 = total_belanja_2029 * 0.10  # Diskon 10%

total_bayar_2029 = total_belanja_2029 - besaran_diskon_2029
rata_harga_2029 = total_belanja_2029 / jumlah_barang_2029
sisa_bagi_2029 = int(total_belanja_2029) % jumlah_barang_2029  # Contoh operator %


# 5. OPERATOR PENUGASAN (Augmented Assignment)
# Menggunakan operator penugasan untuk menambahkan poin bonus jika dapat promo
poin_pelanggan_2029 = 0
if get_promo_2029:
    poin_pelanggan_2029 += 50  # Operator +=


# 6. OPERATOR IDENTITAS (Identity)
# Membuat dua objek untuk mendemonstrasikan perbandingan identitas 'is' dan 'is not'
status_salinan_2029 = status_2029
objek_baru_2029 = str(status_2029)  # Membuat string baru dengan nilai yang sama

cek_identitas_true_2029 = status_salinan_2029 is status_2029
cek_identitas_false_2029 = objek_baru_2029 is not status_2029


# 7. OPERATOR BITWISE
# Representasi biner kondisi pelanggan:
# Bit 0 (0001) -> Member
# Bit 1 (0010) -> Belanja >= 200.000
# Bit 2 (0100) -> Barang >= 3
# Bit 3 (1000) -> Promo Tersedia

bit_member_2029 = 0b0001 if is_member_2029 else 0b0000
bit_belanja_2029 = 0b0010 if syarat_belanja_2029 else 0b0000
bit_barang_2029 = 0b0100 if syarat_barang_2029 else 0b0000
bit_promo_2029 = 0b1000 if promo_tersedia_2029 else 0b0000

# Bitwise OR (|) untuk menggabungkan kondisi
kode_status_2029 = bit_member_2029 | bit_belanja_2029 | bit_barang_2029 | bit_promo_2029

# Bitwise AND (&) untuk cek status tertentu
cek_member_bitwise_2029 = kode_status_2029 & 0b0001
cek_promo_bitwise_2029 = kode_status_2029 & 0b1000

# Bitwise XOR (^) untuk membandingkan dengan kode referensi (misal referensi standar = 1011 (11))
kode_referensi_2029 = 0b1011
hasil_xor_2029 = kode_status_2029 ^ kode_referensi_2029

# Bitwise Shift Left (<<)
hasil_shift_2029 = kode_status_2029 << 1


# ================= OUTPUT PROGRAM =================

print("\n=== DATA TRANSAKSI ===")
print(f"Nama Pelanggan       : {nama_2029}")
print(f"Status Pelanggan     : {status_2029}")
print(f"Total Belanja        : Rp{total_belanja_2029:.0f}")
print(f"Jumlah Barang        : {jumlah_barang_2029}")
print(f"Kode Promo           : {kode_promo_2029}")

print("\n=== HASIL VALIDASI ===")
print(f"Belanja >= Rp200000        : {syarat_belanja_2029}")
print(f"Jumlah Barang >= 3         : {syarat_barang_2029}")
print(f"Status Member              : {is_member_2029}")
print(f"Kode Promo Tersedia        : {promo_tersedia_2029}")
print(f"Mendapatkan Diskon         : {get_diskon_2029}")
print(f"Mendapatkan Promo          : {get_promo_2029}")

print("\n=== HASIL PERHITUNGAN ===")
print(f"Diskon                     : Rp{besaran_diskon_2029:.0f}")
print(f"Total Pembayaran           : Rp{total_bayar_2029:.0f}")
print(f"Rata-rata Harga Barang     : Rp{rata_harga_2029:.2f}")

print("\n=== HAK AKSES PELANGGAN ===")
print(f"Kode Hak Akses             : {bin(kode_status_2029)[2:].zfill(4)}")
print(f"Member Access              : {cek_member_bitwise_2029 > 0}")
print(f"Promo Access               : {cek_promo_bitwise_2029 > 0}")
print(f"Poin Bonus Ditambahkan     : {poin_pelanggan_2029}")

print("\n=== OPERASI BITWISE ===")
print("=== Kode Status Transaksi ===")
print("0001 | 0010 | 0100 | 1000")
print(f"Kode Biner   : {bin(kode_status_2029)[2:].zfill(4)}")
print(f"Kode Desimal : {kode_status_2029}")

print("\n=== Pemeriksaan Status ===")
print("Cek Member (Kode & 0001)")
print(f"Hasil Biner   : {bin(cek_member_bitwise_2029)[2:].zfill(4)}")
print(f"Hasil Desimal : {cek_member_bitwise_2029}")

print("\nCek Promo (Kode & 1000)")
print(f"Hasil Biner   : {bin(cek_promo_bitwise_2029)[2:].zfill(4)}")
print(f"Hasil Desimal : {cek_promo_bitwise_2029}")

print("\n=== Perbandingan Status ===")
print(f"Kode Transaksi : {bin(kode_status_2029)[2:].zfill(4)}")
print(f"Kode Referensi : {bin(kode_referensi_2029)[2:].zfill(4)}")
print(f"Hasil XOR (^)  : {bin(hasil_xor_2029)[2:].zfill(4)} (Desimal: {hasil_xor_2029})")

print("\n=== Shift ===")
print(f"Shift Left (<< 1) Biner   : {bin(hasil_shift_2029)[2:]}")
print(f"Shift Left (<< 1) Desimal : {hasil_shift_2029}")

print("\n=== SELESAI ===")
```

    a. `and`: Menghasilkan `True` hanya jika kedua kondisi bernilai benar. Contoh: Pelanggan wajib berstatus member dan belanja minimal Rp200.000 untuk mendapatkan diskon.
    
    b. `or`: Menghasilkan `True` jika salah satu atau kedua kondisi bernilai benar. Contoh: Kelayakan promo terpenuhi asalkan kuantitas barang mencukupi atau kode promo yang dimasukkan valid.
    
    c. `not`: Digunakan untuk membalikkan nilai kebenaran (`negasi`). Mengubah `True` menjadi `False` dan sebaliknya. Contoh: not `promo_invalid_2029` memastikan program mengunci akses jika kode yang dimasukkan salah.

    b. `not in`: Memeriksa apakah sebuah nilai tidak terdapat di dalam List. Sangat membantu untuk menyaring input kode promo salah secara instan tanpa perlu struktur perulangan manual.

4. **Perbedaan is dan == pada Python:**

    a. `==` (Equality): Membandingkan nilai atau isi dari dua objek. Jika teks atau angkanya sama, hasilnya `True`.
    
    b. `is` (Identity): Membandingkan identitas memori (alamat ID objek) di sistem komputer. Dua objek bisa memiliki teks isi yang persis sama (`==` bernilai `True`), tetapi jika mereka disimpan di lokasi memori yang berbeda, operator is akan menghasilkan `False`.

3. **Operator Logika (`and, or, not`)**

    Bertindak sebagai instrumen penggabung beberapa indikator perbandingan kondisional.

        and: Memastikan sepasang prasyarat wajib dipenuhi secara bersamaan (misalnya berstatus member sekaligus belanja \[\ge \] Rp200.000).
        
        or: Membuka akses kemudahan syarat di mana promo akan aktif jika salah satu parameter terpenuhi.
