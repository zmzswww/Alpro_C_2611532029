## 2.1 Dasar Teori
Praktikum ini melingkupi tiga pilar utama dalam alur kerja rekayasa perangkat lunak modern:

1. **Python**: Bahasa pemrograman tingkat tinggi (high-level programming language) yang bersifat interaktif, dinamis, dan berorientasi objek. Python menggunakan interpreter sehingga eksekusi kode dilakukan baris demi baris, mempermudah proses pengujian fungsi secara langsung.
2. **Visual Studio Code (VS Code)**: Code editor lintas platform yang dikembangkan oleh Microsoft. 3. **Version Control System (Git) & GitHub**: Git adalah perangkat lunak open-source untuk melacak modifikasi berkas teks di mesin lokal secara terdistribusi. Sementara GitHub merupakan layanan berbasis cloud yang meng-host repositori Git, memungkinkan pencadangan kode jarak jauh (remote repository) serta mendukung alur kerja kolaborasi antarpengembang melalui mekanisme commit, push, dan pull.

### 2.2.2 Membuat Repository Github

setelah program dibuat dan dilakukan uji coba pada VS Code, proyek dapat dihubungkan dengan Github. Repository digunakan untuk menyimpan kode program secara daring.

Tahapan yang dilakukan yaitu:

1. Membuat Repository baru pada Github.
2. Membuka folder proyek menggunakan VS Code 
3. Membuka terminal pada VS Code
4. Menginisialisasi Git pada folder proyek
5. Menambahkan file ke dalam staging 
6. Membuat commit
7. Menghubungkan repository lokal dengan repository Github
8. Melakukan push untuk mengunggah kode ke Github.

contoh perintah yang digunakan adalah:

```python
git init
git add .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/nama-repository.git
git push -u origin main
```

Penjelasan:
1. git init digunakan untuk membuat repository Git pada folder proyek.
2. git add . digunakan untuk menambahkan seluruh perubahan ke staging area.
3. git commit digunakan untuk menyimpan perubahan dengan sebuah pesan.
4. git branch -M main digunakan untuk menentukan branch utama dengan nama main.
5. git remote add origin digunakan untuk menghubungkan repository lokal dengan repository GitHub.
6. git push digunakan untuk mengunggah kode dari komputer ke GitHub.
