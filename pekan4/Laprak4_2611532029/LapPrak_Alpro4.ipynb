## 2.1 Dasar Teori
Kondisional merupakan sebuah fitur kontrol alur pada Python yang bertugas mengendalikan eksekusi baris kode berdasarkan terpenuhi atau tidaknya suatu syarat. Lewat mekanisme evaluasi ini, program akan menilai kebenaran dari suatu kondisi (apakah bernilai True atau False) sebelum menentukan langkah pengerjaan selanjutnya.

1. Instruksi `if`
    Digunakan untuk memblokir sebuah perintah agar hanya berjalan jika syarat tunggal yang dipasang terbukti bernilai benar (True).
    
2. Instruksi `if-else`
    Diterapkan apabila sistem dihadapkan pada dua pilihan skenario yang saling bertolak belakang. Bila syarat pertama pada komponen `if` gagal terpenuhi, maka sistem secara otomatis akan melompat ke blok perintah `else`.
    
3. Instruksi `if-elif-else`
    Menjadi solusi ideal ketika parameter yang perlu diverifikasi berjumlah lebih dari dua opsi. Python akan melacak validitas tiap kondisi dari atas ke bawah hingga menemukan satu kecocokan logika yang bernilai benar.

```python
ipk_2029 = float(input("Inputkan IPK anda: "))

if ipk_2029 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_2029))

print("Program Selesai")
```

### 2.2.3 Program "multi_if1_2611532029"

4. Pengujian kesatu: Memeriksa apakah `umur_2029 >= 17` bernilai benar sekaligus status `sim_2029 == "y"`. Jika klop, muncul teks izin berkendara bagi orang dewasa.

6. Pengujian ketiga: Melacak kondisi saat usia di bawah 17 tahun dan tidak memegang SIM. Pesan yang dicetak menyatakan belum cukup umur berkendara.

7. Pengujian keempat: Memvalidasi anomali jika usia belum 17 tetapi menginput sudah ber-SIM, yang akan menghasilkan output ketidakabsahan dokumen SIM.

8. Setelah rangkaian pengecekan selesai disisir satu per satu, program pun berakhir.

### 2.2.4 Program "multi_if2_2611532029"

```python
#input dari user
total_belanja_2029 = float(input("Masukkan total belanja (Rp): "))

#input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_2029 = input("Apakah anda member? (y/t): ").strip().lower()
is_member_2029 = input_member_2029 in ["y", "ya"]

#input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_2029 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_2029 = input_promo_2029 in ["y", "ya"]

total_diskon_persen_2029 = 0

if total_belanja_2029 > 1000000:
    total_diskon_persen_2029 += 10 #Diskon belanja besar

if is_member_2029:
    total_diskon_persen_2029 += 5 #Diskon member

if kode_promo_valid_2029:
    total_diskon_persen_2029 += 15 #Diskon voucher

nominal_diskon_2029 = total_belanja_2029 * (total_diskon_persen_2029 / 100)
total_bayar_2029 = total_belanja_2029 - nominal_diskon_2029

print("\n--- Rincian Pembayaran ---")
print(f"Total diskon    : {total_diskon_persen_2029}% (Rp {nominal_diskon_2029:,.0f})")
print(f"Total bayar     : Rp {total_bayar_2029:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_2029}%")
```

**Ulasan Alur Kode:**

1. Langkah mula dipicu oleh instruksi pengisian nilai nominal uang belanja pembeli lewat fungsi `input()`.

2. Format data angka tersebut dimutasi ke tipe data pecahan (`float`) dan dimuat ke variabel `total_belanja_2029`.

3. Sistem mengonfirmasi keanggotaan pelanggan lewat respons pilihan huruf 'y' atau 't'. Pengetikan distandarisasi lewat metode `.lower()` (mengecilkan huruf) serta `.strip()` (memangkas spasi liar).

4. Hasil validasi string 'y' / 't' dipetakan ke dalam variabel logika `is_member_2029` memakai operator `in`.

5. Prosedur pembersihan teks yang sejenis diterapkan pada kupon promo untuk mengisi isi variabel `kode_promo_valid_2029`.

6. Variabel `total_diskon_persen_2029` disiapkan di awal dengan besaran nilai mula `0`.

7. Filter kesatu mendeteksi nominal transaksi; jika melampaui **Rp1.000.000**, tabungan persen diskon tumbuh **10%**.

8. Filter kedua mengevaluasi kartu keanggotaan; bila terverifikasi aktif, kuota bonus bertambah lagi **5%**.

9. Filter ketiga membaca keabsahan kupon promo; jika terbukti legal, bonus persentase dipertebal **15%**.

10. Berhubung blok kondisional dipisahkan sebagai unit `if` mandiri, persentase potongan dari kriteria berlainan bisa **saling berakumulasi** (ditumpuk).

11. Rumus matematika dijalankan guna mencari tahu besaran rupiah dari kalkulasi total persen diskon terhadap nilai belanja.

12. Nilai bersih pembayaran diperoleh dengan memotong total pengeluaran awal dengan nominal rupiah diskon.

13. Rincian pembukuan dicetak ke layar mencakup besaran persen promo, potongan rupiah, hingga kas bersih yang wajib disetor.

14. Sesi ditutup dengan penayangan rangkuman akumulasi kuota diskon berbentuk persentase.

### 2.2.5 Program "if_elif_else1_2611532029"

4. Penyaringan level satu mengecek apakah umur minimal berada di angka **17 tahun** serta memegang SIM sah (`sim_2029 == "y"`). Jika terpenuhi, izin mengemudi mandiri diberikan.

5. Apabila penyaringan awal tidak lolos, blok `elif` berikutnya mendeteksi jika usia sudah mencukupi tetapi tidak ber-SIM. Hasil keluaran menyatakan larangan membawa motor secara independen.

6. Bila masih meloset, baris `elif` ketiga membaca kemungkinan usia belum legal namun bernilai masukan SIM aktif, memicu keluaran pesan ketidaksesuaian administrasi umur.

7. Di kala tak satu pun prasyarat di atas bernilai benar, instruksi default `else` otomatis mengambil alih untuk mengeluarkan teks dilarang membawa motor bagi anak di bawah umur.

5. Jika gagal lagi, tingkat ketiga (`elif`) menyaring perolehan nilai **mulai dari angka 60** untuk menerbitkan huruf mutu **C**.

6. Saringan tingkat empat (`elif`) memantau pemenuhan kriteria nilai **mulai angka 50** guna melekatkan indeks predikat **D**.

7. Jika capaian nilai berada di bawah semua standar minimal yang dipasang, perintah final `else` merilis mutu terendah yaitu **E**.

8. Usai salah satu indeks mutu berhasil dikeluarkan, operasional program langsung ditutup.

### 2.2.7 Program "match_case1_2611532029"

```python
bulan_2029 = int(input("Masukkan angka bulann (1 - 12): "))

match bulan_2029:
    case 1 :
        print("Januari")
    case 2 :
            print("Februari")
    case 3 :
            print("Maret")
    case 4 :
            print("April")
    case 5 :
            print("Mei")
    case 6 :
            print("Juni")
    case 7 :
            print("Juli")
    case 8 :
            print("Agustus")
    case 9 :
            print("September")
    case 10 :
            print("Oktober")
    case 11 :
            print("November")
    case 12 :
            print("Desember")
    case _:
            print("Angka tidak valid")
```

**Ulasan Alur Kode:**

1. Awal pengerjaan menuntun user menginput perwakilan digit urutan bulan dalam rentang batas **1 hingga 12**.

2. Elemen string masukan tersebut diubah ke bilangan bulat (`int`) demi mengisi slot variabel data `bulan_2029`.

3. Skrip memanggil fungsi kontrol struktural **`match-case`** untuk menguji kecocokan isi variabel `bulan_2029` dengan pola kasus yang telah disediakan.

4. Bila pola bernilai cocok dengan angka **1**, string nama bulan **Januari** dipanggil ke layar.

5. Jalur kecocokan berlanjut mengikuti pola angka **2 hingga 12** guna memunculkan penamaan kalender dari Februari sampai dengan Desember.

6. Bila angka masukan melenceng di luar sekat bilangan **1 sampai 12**, tanda pola universal `case _` (wildcard) mengambil alih untuk menerbitkan notifikasi "Angka tidak valid".

7. Eksekusi program dianggap rampung setelah sistem berhasil mengeluarkan salah satu luaran dari kasus di atas.

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")
nama_2029 = input("Masukkan Nama Pengunjung        : ")
umur_2029 = int(input("Input umur anda                 : "))

# .strip().lower() untuk menangani input string
sim_2029 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()

# Menu Paket Wahana
print("\nPilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")

paket_2029 = input("Masukkan nomor paket (1-5)      : ").strip()
jumlah_tiket_2029 = int(input("Masukkan jumlah tiket           : "))

# Pilar 1: IF Tunggal untuk Validasi Kelogisan Tiket
if jumlah_tiket_2029 <= 0:
    print("[PERINGATAN] Kuota tiket tidak valid!")
    exit()

is_member_2029 = input("Apakah Anda member? (y/t)       : ").strip().lower()
kode_promo_valid_2029 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()

# Inisialisasi variabel pendukung
harga_satuan_2029 = 0
status_akses_2029 = ""

# Pilar 2: MATCH - CASE untuk Pemilihan Wahana
match paket_2029:
    case "1":
        harga_satuan_2029 = 50000
    case "2":
        harga_satuan_2029 = 75000
    case "3":
        harga_satuan_2029 = 120000
    case "4":
        harga_satuan_2029 = 100000
    case "5":
        harga_satuan_2029 = 220000
    case _:
        print("Paket wahana tidak valid!")
        exit()

# Pilar 3: IF - ELIF - ELSE dan Operator Logika untuk Validasi Izin Kendali
print("\n--- KELAYAKAN PENGENDARA WAHANA ---")
if paket_2029 == "3":
    if umur_2029 >= 17 and sim_2029 == 'y':
        status_akses_2029 = "Anda sudah dewasa dan boleh mengendarai ATV sendiri."
    elif umur_2029 >= 17 and sim_2029 != 'y':
        status_akses_2029 = "Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur)."
    elif umur_2029 < 17 and sim_2029 == 'y':
        status_akses_2029 = "Identitas tidak valid: Belum cukup umur memiliki SIM."
    else:
        status_akses_2029 = "Anda belum cukup umur dan tidak boleh bawa motor ATV."
else:
    if umur_2029 >= 10:
        status_akses_2029 = "Anda memenuhi syarat umur untuk menikmati wahana ini."
    else:
        status_akses_2029 = "Anda belum cukup umur untuk menikmati wahana ini."

print(f"Status Akses: {status_akses_2029}")

# Perhitungan Subtotal
subtotal_2029 = harga_satuan_2029 * jumlah_tiket_2029
total_diskon_persen_2029 = 0

# Pilar 4: MULTI-IF Terpisah untuk Akumulasi Diskon Bertingkat
if subtotal_2029 >= 200000:
    total_diskon_persen_2029 += 10

if is_member_2029 in ['y', 'ya']:
    total_diskon_persen_2029 += 5

if kode_promo_valid_2029 in ['y', 'ya']:
    total_diskon_persen_2029 += 15

if jumlah_tiket_2029 >= 5:
    total_diskon_persen_2029 += 5

# Perhitungan Nominal Diskon dan Total Bayar
nominal_diskon_2029 = subtotal_2029 * (total_diskon_persen_2029 / 100)
total_bayar_2029 = subtotal_2029 - nominal_diskon_2029

# Pilar 5: IF - ELSE untuk Evaluasi Kelulusan Audit & Bonus
catatan_layanan_2029 = ""
if total_bayar_2029 > 300000:
    catatan_layanan_2029 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
else:
    catatan_layanan_2029 = "Terima kasih telah berkunjung."

# OUTPUT RINCIAN PEMBAYARAN (Format Desimal f-string)
print("\n--- Rincian Pembayaran ---")
print(f"Subtotal Belanja : Rp {subtotal_2029:,.0f}")
print(f"Total Diskon     : {total_diskon_persen_2029}% (Rp {nominal_diskon_2029:,.0f})")
print(f"Total Bayar      : Rp {total_bayar_2029:,.0f}")
print(f"Catatan Layanan  : {catatan_layanan_2029}")
print("Program Selesai")
```

**Penjelasan Tugas**

1. **Input Data & Pembersihan String:** Program meminta data pengunjung seperti nama, umur, status SIM, paket, dan jumlah tiket. Fungsi `.strip().lower()` digunakan untuk memastikan input konfirmasi (seperti 'y' atau 't') tetap valid meskipun ditulis dengan huruf besar atau spasi tambahan.

2. **Validasi Kuota Tiket (`if` Tunggal):** Program memeriksa apakah jumlah tiket bernilai `0` atau `negatif`. Jika ya, program langsung memicu peringatan kuota tidak valid dan menghentikan seluruh proses transaksi secara paksa menggunakan `exit()`.

3. **Pemilihan Paket (`match-case`):** Struktur ini menentukan harga tiket berdasarkan nomor paket (1 s.d. 5). Jika pengguna memasukkan nomor di luar pilihan tersebut, blok default `case _` akan aktif untuk membatalkan transaksi karena paket tidak valid.

4. **Cek Kelayakan (`if-elif-else` & Operator Logika):** Untuk Paket 3 (ATV), operator `and` dan `!=` digunakan untuk memeriksa kombinasi umur dan kepemilikan SIM secara ketat. Sedangkan untuk paket lainnya, sistem beralih ke blok else untuk mengecek syarat umur minimal 10 tahun.

5. **Akumulasi Potongan Harga (`multi-if` Terpisah):** Menggunakan rangkaian blok `if` mandiri (bukan `elif`) agar pemeriksaan berjalan satu per satu dari atas ke bawah. Hal ini memungkinkan diskon belanja besar, status member, kode voucher, dan jumlah rombongan didapatkan secara bertumpuk (akumulatif).

6. **Audit Bonus & Cetak Struk (`if-else`):** Program mengevaluasi total bayar akhir; jika di atas Rp 300.000, pembeli mendapat bonus souvenir gratis. Terakhir, semua rincian biaya dicetak rapi menggunakan f-string format `:,.0f` untuk memunculkan pemisah ribuan.
