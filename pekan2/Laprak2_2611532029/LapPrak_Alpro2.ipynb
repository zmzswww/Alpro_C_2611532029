<h1 align="center">BAB II<br>PEMBAHASAN</h1> 

## 2.1 Dasar Teori
A. Pengertian Variabel
Variabel adalah sebuah lokasi atau wadah di dalam memori komputer yang diberi nama (identifikator) dan digunakan untuk menyimpan suatu nilai data secara dinamis. Nilai di dalam variabel dapat berubah-ubah selama program dieksekusi. Aturan penulisan variabel biasanya berbeda pada setiap bahasa pemrograman, namun secara umum tidak boleh diawali dengan angka dan tidak boleh mengandung spasi.

B. Pengertian Tipe Data
Tipe data adalah klasifikasi atau jenis nilai yang dapat ditampung oleh suatu variabel. Tipe data memberi tahu compiler atau interpreter mengenai cara memperlakukan nilai tersebut (misalnya, apakah nilai tersebut bisa dijumlahkan secara matematis atau hanya bisa digabungkan teksnya).

Berdasarkan struktur praktikum ini, terdapat 5 poin konsep dasar yang dipelajari:
1. Numeric (Numerik): Tipe data standar yang digunakan untuk menyimpan nilai berupa angka. 
Tipe data ini dibagi lagi menjadi dua jenis utama, yaitu Integer untuk bilangan bulat (contoh: 10, -5) dan Float/Real untuk bilangan pecahan atau desimal (contoh: 3.14, 0.5).

2. String (Teks): Tipe data bentukan/komposit yang digunakan untuk menyimpan kumpulan karakter, kata, atau kalimat. 
Penulisannya dalam kode program biasanya diapit oleh tanda kutip (contoh: "Halo Dunia").

3. Boolean (Logika): Tipe data dasar yang hanya memiliki dua kemungkinan nilai, yaitu True (Benar) atau False (Salah). 
Tipe data ini sangat krusial dalam menentukan alur logika dan percabangan program.

4. Input (Masukan Data): Sebuah mekanisme atau fungsi yang digunakan untuk menerima data dari pengguna secara interaktif melalui perangkat masukan (seperti keyboard), yang kemudian nilainya akan disimpan ke dalam variabel dengan tipe data tertentu.

5. Konstanta: Sebuah entitas penyimpanan nilai di dalam memori yang mirip dengan variabel, namun memiliki karakteristik mutlak di mana nilai yang telah diisi di awal program bersifat tetap dan tidak dapat diubah-ubah sepanjang program dieksekusi.

### 2.2.1 Program Numeric

```python
# Buat file dengan nama Numerik_NIM.py
# nama variabel ditambah 4 digit terakhir NIM contoh: a_2029
#create a variable with numeric type value.
a_2029=100
print("The type of variable having value", a_2029, " is ", type (a_2029))

#create a variable with float value.
b_2029=10.2345
print("The type of variable having value", b_2029, " is ", type (b_2029))

#create a variable with complex value.
c_2029=100+3j
print("The type of variable having value", c_2029, " is ", type (c_2029))
```

**Penjelasan :**
1. Deklarasi a_2029 dengan nilai 100 sebagai integer.
2. Cetak ("The type of variable having value",a_2029,"is",type(a_2029)) dengan type() digunakan untuk mengetahui tipe variabel dari a_2029.
3. Deklarasi b_2029 dengan nilai 10.2345 sebagai float.
4. Cetak ("The type of variable having value",b_2029,"is",type(b_2029)) dengan type() digunakan untuk mengetahui tipe variabel dari b_2029.
5. Deklarasi c_2029 dengan nilai (100 + 3j) sebagai komplek.
6. Cetak ("The type of variable having value",c_2029,"is",type(c_2029)) dengan type() digunakan untuk mengetahui tipe variabel dari c_2029

### 2.2.3 Program Boolean

```python
# Buat file dengan nama Boolean_NIM.py
# nama variabel ditambah 4 digit terakhir NIM contoh: nilai_2029
# Deklarasi variabel dengan tipe Boolean
is_lulus_2029 = True
is_cumlaude_2029 = True

# Menggunakan Boolean
nilai_2029 = 85
batas_lulus_2029 = 75

# Menentukan nilai Boolean dari kondisi
status_kelulusan_2029 = nilai_2029 >= batas_lulus_2029 #Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai: ", nilai_2029)
print("Apakah Lulus?: ", status_kelulusan_2029)
if is_lulus_2029 and is_cumlaude_2029:
    print("Selamat, Anda lulus dengan predikat Cumlaude!")
```

Penjelasan:
1. Deklarasi is_lulus_2029 = True sebagai boolean
2. Deklarasi is_cumlaude_2029 = True sebagai boolean
3. Deklarasi nilai_2029 = 85 sebagai integer
5. Deklarasi batas_lulus_2029 = 75 sebagai integer
6. Deklarasi status_kelulusan_2029 = nilai_2029 >= batas_lulus_2029 sebagai boolean
7. Cetak ("=== Check Kelulusan ===")
8. Cetak ("Nilai: ",nilai_2029)
9. Cetak ("Apakah lulus?: ",status_kelulusan_2029)
10. Jika is_lulus_2029 dan is_cumlaude_2029 bernilai benar cetak ("Selamat, Anda lulus dengan predikat Cumlaude!")

### 2.2.4 Program Input

```python
# Buat file dengan nama Input_NIM.py
# Buat program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit nimm terakhir contoh: panjang_2029
# Program ini menggunakan fungsi input() untuk meminta pengguna
# Memasukkan nilai panjang dan lebar persegi panjang
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_2029 = input('Masukkan nilai panjang: ')
lebar_2029 = input('Masukkan nilai lebar: ')
luas_2029 = int(panjang_2029) * int(lebar_2029)
print("Luas= ", luas_2029)
```

### 2.2.5 Program Konstanta

```python
# Buat file dengan nama Konstanta_2611532029
# Program ini menggunakan konstanta untuk menghitun luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_2029
from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_2029 = float(input('Masukkan nilai jari-jari: '))
luas_2029 = PI * jari_2029 * jari_2029
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2029, luas_2029))
```

**Penjelasan:**
1. Pemanggilan Final dari library typing from typing import Final
2. Deklarasi PI: Final = 3.14 sebagai konstanta
3. Cetak ("Pi: %f" % (PI))
4. Nilai jari_2029 = float(input("Masukkan nilai jari-jari: "))
5. Nilai luas_2029 = PI * jari_2029
6. Cetak ("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2029,luas_2029))
